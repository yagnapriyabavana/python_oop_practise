```python
class Student:
    def __init__(self,name,marks,age):
        self.name=name
        self.marks=marks
        self.age=age
    def details(self):
        print(f'Sudent name :{self.name}')
        print(f'Student marks: {self.marks}')
        print(f'Student age: {self.age}')
    def avg_marks(self):
        count=0
        for x in self.marks:
            count+=x
        print(f'the avg marks is {count/len(self.marks)}')
    def calculate_percentage(self):
        count=0
        for x in self.marks:
            count+=x
        self.per=(count/90)*100
        print(f'The percentage is {self.per}')
    def cal_grade(self):
        if self.per>= 90:
            print("A+")
        elif self.per>=80 and self.per<=89:
            print("A")
        elif self.per>=70 and self.per<=79:
            print("B")
        else:
            print("C")
s1=Student('Ram',[24,27,30],21)
s1.details()
s1.avg_marks()
s1.calculate_percentage()
s1.cal_grade()
```

```python
class Employee:
    def __init__(self,name,employee_id,salary):
        self.name=name
        self.employee_id=employee_id
        self.salary=salary
    def display_details(self):
        print(f'Employee name :{self.name}')
        print(f'Employee id : {self.employee_id}')
        print(f'Employee salary :{self.salary}')
class Developer(Employee):
    def __init__(self,name,employee_id,salary,programming_language):
        super().__init__(name,employee_id,salary)
        self.programming_language=programming_language
    def display_language(self):
        print(f'Programming language is: {self.programming_language}')
class Manager(Employee):
    def __init__(self,name,employee_id,salary,team_size):
        super().__init__(name,employee_id,salary)
        self.team_size=team_size
    def display_team(self):
        print(f'Team size:{self.team_size}')
d1=Developer('Ram',101,50000,'Python')
d1.display_details()
d1.display_language()
m1=Manager('Priya',102,70000,5)
m1.display_details()
m1.display_team()
```

```python
class Payment:
    def __init__(self,amount):
        self.amount=amount
class CreditCard(Payment):
    def pay(self):
        print(f'Credit card amount:{self.amount}')
class UPI(Payment):
    def pay(self):
        print(f'UPI payment of:{self.amount}')
class Cash(Payment):
    def pay(self):
        print(f'Cash payment is:{self.amount}')
c1=CreditCard(2000)
u1=UPI(1500)
c=Cash(500)
for x in [c1,u1,c]:
    x.pay()
```

```python
class BankAccount:
    def  __init__(self,name,balance):
        self.name=name
        self.__balance=balance
    def details(self):
        print("Account holder:",self.name)
    def balance(self):
        return self.__balance
    def deposit(self,amount):
        self.amount=amount
        if self.amount<=0:
            print("Negative Deposits not allowed")
        else:
            print(f'Deposited money  :{self.amount}')
            self.__balance+=self.amount
    def withdraw(self,amount):
        self.amount=amount
        if self.amount > self.__balance:
            print("Withdraw greater than balance not accepted")
        else:
            print(f'Withdraw money : {self.amount}')
            self.__balance-=self.amount
    def get_balance(self):
        return self.__balance
b=BankAccount('Ram',5000)
b.details()
print(b.balance())
b.deposit(1000)
b.withdraw(2000)
print(b.get_balance())
```
